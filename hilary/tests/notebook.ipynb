```python
"""Execute hilary with command line."""
from __future__ import annotations

from multiprocessing import cpu_count
from pathlib import Path

import typer

from hilary.inference_new import HILARy
from hilary.utils import (
    create_classes,
    get_logger,
    pairwise_evaluation,
    preprocess,
    read_input,
    save_dataframe,
)
```

```python
data_path=Path("/home/athenes/gitlab/HILARy/316188_small.tsv")
config=None
```

```python
dataframe = read_input(input_path=data_path, config=config)
dataframe["sequence_id"] = dataframe.index.astype("str")
dataframe["sequence_id"]=dataframe["sequence_id"].astype(str)
dataframe["sequence_id"] = dataframe["sequence_id"].str.strip("-igh")
dataframe.set_index("sequence_id")
paired = False
```

```python
dataframe_processed = preprocess(df=dataframe)
classes = create_classes(dataframe_processed)
hilary = HILARy(
    df=dataframe_processed,
    classes = classes,
    paired=paired,
)
```

```python
from itertools import combinations
from textdistance import hamming, levenshtein
for _, group in dataframe_processed.groupby(["v_gene", "j_gene", "cdr3_length"]):
    alignments = group["alt_germline_alignment"].dropna().tolist()
    if len(group["alt_germline_alignment"].str.len().unique())==1:
        continue
    for u, v in combinations(alignments, 2):
        # Do something with each unique pair u, v
        print(len(u), len(v), levenshtein(u,v))
```

```python
dataframe["sequence_id"] = dataframe["sequence_id"] + "-igh"
hilary.get_xy_thresholds(df=dataframe_processed)
```

```python
dataframe_inferred = hilary.infer(df=dataframe_processed)
dataframe["clone_id"] = dataframe_inferred["clone_id"]
```
